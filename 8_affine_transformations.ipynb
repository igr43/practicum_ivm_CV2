```python
import cv2
def resize_with_aspect_ratio(image, width=None, height=None, inter=cv2.INTER_AREA):
    dim = None
    (h, w) = image.shape[:2]
    
    if width is None and height is None:
        return image
        
    if width is None:
        r = height / float(h)
        dim = (int(w * r), height)
    else:
        r = width / float(w)
        dim = (width, int(h * r))
        
    return cv2.resize(image, dim, interpolation=inter)
```

```python
import cv2
import numpy as np

# Загружаем изображение
img = cv2.imread('images/laptop.jpg')

# Исходные точки (например, углы треугольника)
src_pts = np.float32([[50, 50], [200, 50], [50, 200]])

# Новые позиции этих точек
dst_pts = np.float32([[10, 100], [200, 50], [100, 250]])

# Получаем матрицу преобразования
M = cv2.getAffineTransform(src_pts, dst_pts)

# Применяем преобразование
result = cv2.warpAffine(img, M, (img.shape[1], img.shape[0]))

# Показываем результат
cv2.imshow("Affine Transform", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# Загружаем изображение
img = cv2.imread('images/book.png')

# Размер изображения
rows, cols = img.shape[:2]

# Четыре угла документа на исходном изображении
src_points = np.float32([[100, 200], [400, 180], [120, 500], [420, 520]])

# Желаемое прямоугольное отображение
dst_points = np.float32([[0, 0], [300, 0], [0, 400], [300, 400]])

# Получаем матрицу перспективной трансформации
M = cv2.getPerspectiveTransform(src_points, dst_points)

# Применяем трансформацию
warped = cv2.warpPerspective(img, M, (300, 400))

# Показываем результат
cv2.imshow("Original", img)
cv2.imshow("Warped", warped)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

points = []

def click_event(event, x, y, flags, param):
    global points, img
    
    if event == cv2.EVENT_LBUTTONDOWN and len(points) < 4:
        points.append([x, y])
        cv2.circle(img, (x, y), 5, (0, 0, 255), -1)
        cv2.imshow("image", img)
        print(f"Точка {len(points)}: ({x}, {y})")

# Загрузка изображения
img = cv2.imread("images/book.png")
img = resize_with_aspect_ratio(img, width=500) 
cv2.imshow("image", img)
cv2.setMouseCallback("image", click_event)

print("Кликните по 4 углам стола")
print("Порядок: верхний левый, верхний правый, нижний левый, нижний правый")

cv2.waitKey(0)
cv2.destroyAllWindows()

# Перспективная трансформация
if len(points) == 4:
    src = np.float32(points)
    
    # Вычисляем размер
    width = int(max(
        np.linalg.norm(src[1] - src[0]),
        np.linalg.norm(src[3] - src[2])
    ))
    
    height = int(max(
        np.linalg.norm(src[2] - src[0]),
        np.linalg.norm(src[3] - src[1])
    ))
    
    # Целевые точки
    dst = np.float32([[0, 0], [width, 0], [0, height], [width, height]])
    
    # Трансформация
    M = cv2.getPerspectiveTransform(src, dst)
    result = cv2.warpPerspective(img, M, (width, height))
    
    cv2.imshow("result", result)
    cv2.imwrite("images/book_topdown.jpg", result)
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Нужно выбрать 4 точки!")
```
