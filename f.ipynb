```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle, Ellipse, FancyArrowPatch, Polygon
import matplotlib.patches as mpatches

def create_clean_figure_1_8():
    """Чистая схема для задачи 1.8 без текста на графике"""
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Настройки графика
    ax.set_xlim(-2.2, 2.2)
    ax.set_ylim(-2.2, 2.2)
    ax.set_aspect('equal')
    ax.axis('off')  # Отключаем оси
    
    # Рисуем реку (голубой прямоугольник)
    river = Rectangle((-2, -2), 4, 4, 
                     facecolor='lightblue', alpha=0.2, 
                     edgecolor='blue', linewidth=2)
    ax.add_patch(river)
    
    # Бакен в центре (чёрная точка с жёлтой серединой)
    ax.plot(0, 0, 'ko', markersize=15, markeredgewidth=2)
    ax.plot(0, 0, 'yo', markersize=7)
    
    # Стрелка течения реки (синяя)
    ax.arrow(-1.5, 1.8, 3.0, 0, 
             head_width=0.1, head_length=0.2, 
             fc='blue', ec='blue', width=0.03, 
             linewidth=2, alpha=0.7)
    
    # Система координат (чёрные стрелки)
    ax.arrow(0, 0, 1.8, 0, 
             head_width=0.08, head_length=0.18, 
             fc='black', ec='black', width=0.02, 
             linewidth=1.5)
    ax.arrow(0, 0, 0, 1.8, 
             head_width=0.08, head_length=0.18, 
             fc='black', ec='black', width=0.02, 
             linewidth=1.5)
    
    # Лодка A - движение вдоль реки (красные стрелки)
    ax.arrow(0, 0.4, 1.6, 0, 
             head_width=0.08, head_length=0.16, 
             fc='red', ec='red', width=0.02, 
             linewidth=2.5)
    ax.arrow(0, 0.4, -1.6, 0, 
             head_width=0.08, head_length=0.16, 
             fc='red', ec='red', width=0.02, 
             linewidth=2.5, linestyle='--', alpha=0.7)
    
    # Лодка B - движение поперёк реки (зелёные стрелки)
    ax.arrow(0.4, 0, 0, 1.6, 
             head_width=0.08, head_length=0.16, 
             fc='green', ec='green', width=0.02, 
             linewidth=2.5)
    ax.arrow(0.4, 0, 0, -1.6, 
             head_width=0.08, head_length=0.16, 
             fc='green', ec='green', width=0.02, 
             linewidth=2.5, linestyle='--', alpha=0.7)
    
    # Метки точек возврата
    ax.plot(1.6, 0.4, 'ro', markersize=8)
    ax.plot(-1.6, 0.4, 'ro', markersize=8)
    ax.plot(0.4, 1.6, 'go', markersize=8)
    ax.plot(0.4, -1.6, 'go', markersize=8)
    
    plt.tight_layout()
    plt.savefig('figure_1_8_clean.png', dpi=300, bbox_inches='tight', pad_inches=0.1)
    plt.show()

def create_clean_figure_1_95():
    """Чистая схема для задачи 1.95 без текста на графике"""
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Параметры эллипса
    a = 2.0  # большая полуось
    b = 1.2  # малая полуось
    omega = 2.0
    
    # Настройки графика
    ax.set_xlim(-2.5, 2.5)
    ax.set_ylim(-1.8, 1.8)
    ax.set_aspect('equal')
    ax.axis('off')  # Отключаем оси
    
    # Рисуем эллиптическую траекторию
    t = np.linspace(0, 2*np.pi, 1000)
    x = a * np.sin(omega * t)
    y = b * np.cos(omega * t)
    ax.plot(x, y, 'b-', linewidth=1.5, alpha=0.6)
    
    # Начало координат
    ax.plot(0, 0, 'ko', markersize=10, markeredgewidth=2)
    
    # Выбираем точку на траектории
    t0 = np.pi/4  # 45 градусов
    x0 = a * np.sin(omega * t0)
    y0 = b * np.cos(omega * t0)
    
    # Рисуем частицу в выбранный момент
    ax.plot(x0, y0, 'ro', markersize=10, markeredgewidth=2)
    
    # Радиус-вектор (зелёный)
    ax.arrow(0, 0, x0, y0, 
             head_width=0.08, head_length=0.15, 
             fc='green', ec='green', width=0.01, 
             linewidth=2)
    
    # Скорость (красный)
    vx0 = a * omega * np.cos(omega * t0)
    vy0 = -b * omega * np.sin(omega * t0)
    scale_v = 0.4
    ax.arrow(x0, y0, scale_v * vx0, scale_v * vy0, 
             head_width=0.06, head_length=0.12, 
             fc='red', ec='red', width=0.008, 
             linewidth=2)
    
    # Ускорение/Сила (фиолетовый)
    ax0 = -a * omega**2 * np.sin(omega * t0)
    ay0 = -b * omega**2 * np.cos(omega * t0)
    scale_a = 0.3
    ax.arrow(x0, y0, scale_a * ax0, scale_a * ay0, 
             head_width=0.06, head_length=0.12, 
             fc='purple', ec='purple', width=0.008, 
             linewidth=2)
    
    # Показываем, что ускорение направлено к центру
    ax.plot([x0, 0], [y0, 0], 'k--', alpha=0.3)
    
    # Координатные оси
    ax.arrow(-2.3, 0, 4.6, 0, 
             head_width=0.08, head_length=0.15, 
             fc='black', ec='black', width=0.01, 
             linewidth=1, alpha=0.5)
    ax.arrow(0, -1.6, 0, 3.2, 
             head_width=0.08, head_length=0.15, 
             fc='black', ec='black', width=0.01, 
             linewidth=1, alpha=0.5)
    
    plt.tight_layout()
    plt.savefig('figure_1_95_clean.png', dpi=300, bbox_inches='tight', pad_inches=0.1)
    plt.show()

def create_velocity_diagram_1_8():
    """Диаграмма скоростей для задачи 1.8"""
    fig, ax = plt.subplots(figsize=(6, 6))
    
    # Настройки графика
    ax.set_xlim(-0.2, 2.2)
    ax.set_ylim(-0.2, 2.2)
    ax.set_aspect('equal')
    ax.axis('off')  # Отключаем оси
    
    # Параметры
    u = 1.0      # скорость течения
    eta = 1.2    # отношение скоростей
    v = eta * u  # скорость лодки относительно воды
    v_y = np.sqrt(v**2 - u**2)  # поперечная составляющая
    
    # Начало координат
    ax.plot(0, 0, 'ko', markersize=8)
    
    # Вектор скорости течения (горизонтальный)
    ax.arrow(0, 0, u, 0, 
             head_width=0.07, head_length=0.14, 
             fc='blue', ec='blue', width=0.015, 
             linewidth=2)
    
    # Поперечная составляющая скорости (вертикальная)
    ax.arrow(0, 0, 0, v_y, 
             head_width=0.07, head_length=0.14, 
             fc='green', ec='green', width=0.015, 
             linewidth=2)
    
    # Вектор скорости лодки относительно воды (гипотенуза)
    ax.arrow(0, 0, u, v_y, 
             head_width=0.07, head_length=0.14, 
             fc='red', ec='red', width=0.015, 
             linewidth=2)
    
    # Пунктирные линии для прямоугольного треугольника
    ax.plot([u, u], [0, v_y], 'k--', alpha=0.5, linewidth=1)
    ax.plot([0, u], [v_y, v_y], 'k--', alpha=0.5, linewidth=1)
    
    # Прямой угол в начале координат
    angle_box = Polygon([[0, 0], [0.2, 0], [0, 0.2]], 
                       closed=True, fill=True, 
                       facecolor='black', alpha=0.3)
    ax.add_patch(angle_box)
    
    plt.tight_layout()
    plt.savefig('velocity_diagram_clean.png', dpi=300, bbox_inches='tight', pad_inches=0.1)
    plt.show()

def create_all_figures():
    """Создание всех трёх рисунков"""
    print("Создание чистых рисунков для задач...")
    
    create_clean_figure_1_8()
    print("✓ Создан figure_1_8_clean.png")
    
    create_clean_figure_1_95()
    print("✓ Создан figure_1_95_clean.png")
    
    create_velocity_diagram_1_8()
    print("✓ Создан velocity_diagram_clean.png")
    
    print("\nВсе рисунки сохранены в текущей директории.")
    print("Используйте их в LaTeX документе следующим образом:")
    print()
    print("\\begin{figure}[h]")
    print("\\centering")
    print("\\includegraphics[width=0.6\\textwidth]{figure_1_8_clean.png}")
    print("\\caption{Схема движения лодок на реке}")
    print("\\label{fig:boat_motion}")
    print("\\end{figure}")

if __name__ == "__main__":
    create_all_figures()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch, Circle, Ellipse
from matplotlib.patches import Rectangle, Polygon
import matplotlib.patches as mpatches

def create_figure_1_8():
    """Создание рисунка для задачи 1.8"""
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # Левая панель: схема реки
    ax1.set_xlim(-2, 2)
    ax1.set_ylim(-2, 2)
    ax1.set_aspect('equal')
    ax1.grid(True, alpha=0.3)
    ax1.set_title('Схема реки и движения лодок')
    
    # Река (прямоугольник)
    river = Rectangle((-1.8, -1.8), 3.6, 3.6, 
                      facecolor='lightblue', alpha=0.3, edgecolor='blue')
    ax1.add_patch(river)
    
    # Бакен
    ax1.plot(0, 0, 'ko', markersize=10, label='Бакен')
    ax1.plot(0, 0, 'yo', markersize=5)
    
    # Стрелка течения реки
    ax1.arrow(-1.5, 1.7, 2.8, 0, head_width=0.1, head_length=0.15, 
              fc='blue', ec='blue', width=0.02, label='Течение (u)')
    
    # Система координат
    ax1.arrow(0, 0, 1, 0, head_width=0.05, head_length=0.1, 
              fc='black', ec='black', width=0.01)
    ax1.arrow(0, 0, 0, 1, head_width=0.05, head_length=0.1, 
              fc='black', ec='black', width=0.01)
    ax1.text(1.05, 0, 'x (вдоль реки)', ha='left', va='center')
    ax1.text(0, 1.05, 'y (поперёк)', ha='center', va='bottom')
    
    # Траектории лодок
    # Лодка A (вдоль реки)
    ax1.arrow(0, 0, 1.2, 0, head_width=0.05, head_length=0.1, 
              fc='red', ec='red', width=0.015, label='Лодка A')
    ax1.arrow(0, -0.1, -1.2, 0, head_width=0.05, head_length=0.1, 
              fc='red', ec='red', width=0.015)
    ax1.text(1.2, 0, 'A', ha='left', va='center', color='red', fontweight='bold')
    
    # Лодка B (поперёк реки)
    ax1.arrow(0, 0, 0, 1.2, head_width=0.05, head_length=0.1, 
              fc='green', ec='green', width=0.015, label='Лодка B')
    ax1.arrow(0.1, 0, 0, -1.2, head_width=0.05, head_length=0.1, 
              fc='green', ec='green', width=0.015)
    ax1.text(0, 1.2, 'B', ha='center', va='bottom', color='green', fontweight='bold')
    
    # Правая панель: диаграмма скоростей для лодки B
    ax2.set_xlim(-0.5, 2.5)
    ax2.set_ylim(-0.5, 2.5)
    ax2.set_aspect('equal')
    ax2.grid(True, alpha=0.3)
    ax2.set_title('Диаграмма скоростей для лодки B')
    
    # Начало координат
    ax2.plot(0, 0, 'ko', markersize=8)
    
    # Вектор скорости течения
    ax2.arrow(0, 0, 1, 0, head_width=0.1, head_length=0.15, 
              fc='blue', ec='blue', width=0.02, label='Течение (u)')
    ax2.text(1.1, 0, 'u', ha='left', va='center', color='blue')
    
    # Вектор скорости лодки относительно воды
    v = 1.2  # η = 1.2
    v_y = np.sqrt(v**2 - 1**2)  # v_y = sqrt(v^2 - u^2), u=1
    
    ax2.arrow(0, 0, 0, v_y, head_width=0.1, head_length=0.15, 
              fc='green', ec='green', width=0.02, label='v_y (поперёк)')
    ax2.text(0.2, v_y/2, f'v_y = √({v}² - 1²)', ha='left', va='center', color='green')
    
    # Вектор скорости лодки относительно воды (результирующий)
    ax2.arrow(0, 0, 1, v_y, head_width=0.1, head_length=0.15, 
              fc='red', ec='red', width=0.02, label='v (отн. воды)')
    ax2.text(1.1, v_y/2, f'v = ηu = {v}', ha='left', va='center', color='red')
    
    # Пунктирные линии для прямоугольного треугольника
    ax2.plot([1, 1], [0, v_y], 'k--', alpha=0.5)
    ax2.plot([0, 1], [v_y, v_y], 'k--', alpha=0.5)
    
    # Угол
    angle = np.arctan(v_y/1)
    arc = mpatches.Arc((0, 0), 0.8, 0.8, angle=0, theta1=0, theta2=angle*180/np.pi)
    ax2.add_patch(arc)
    ax2.text(0.3, 0.1, 'α', ha='center', va='center')
    
    ax2.legend(loc='upper right')
    ax1.legend(loc='upper right')
    
    plt.tight_layout()
    plt.savefig('problem_1_8.png', dpi=300, bbox_inches='tight')
    plt.show()

def create_figure_1_95():
    """Создание рисунка для задачи 1.95"""
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # Параметры
    a = 2.0  # полуось по x
    b = 1.0  # полуось по y
    omega = 2.0  # угловая частота
    
    # Левая панель: траектория и векторы в момент времени t
    ax1.set_xlim(-2.5, 2.5)
    ax1.set_ylim(-1.5, 1.5)
    ax1.set_aspect('equal')
    ax1.grid(True, alpha=0.3)
    ax1.set_title('Движение частицы по эллиптической траектории')
    
    # Эллиптическая траектория
    t = np.linspace(0, 2*np.pi, 1000)
    x = a * np.sin(omega * t)
    y = b * np.cos(omega * t)
    ax1.plot(x, y, 'b-', alpha=0.5, label='Траектория (эллипс)')
    
    # Выбираем момент времени для отображения векторов
    t0 = np.pi/6  # 30 градусов
    x0 = a * np.sin(omega * t0)
    y0 = b * np.cos(omega * t0)
    
    # Радиус-вектор
    ax1.arrow(0, 0, x0, y0, head_width=0.1, head_length=0.15, 
              fc='green', ec='green', width=0.01, label='Радиус-вектор r')
    ax1.text(x0/2, y0/2, 'r', ha='center', va='center', color='green', fontsize=12)
    
    # Скорость
    vx0 = a * omega * np.cos(omega * t0)
    vy0 = -b * omega * np.sin(omega * t0)
    scale_v = 0.5  # масштаб для наглядности
    ax1.arrow(x0, y0, scale_v*vx0, scale_v*vy0, head_width=0.05, head_length=0.1, 
              fc='red', ec='red', width=0.008, label='Скорость v')
    ax1.text(x0 + scale_v*vx0/2, y0 + scale_v*vy0/2, 'v', 
             ha='center', va='center', color='red', fontsize=12)
    
    # Ускорение (сила, делённая на m)
    ax0 = -a * omega**2 * np.sin(omega * t0)
    ay0 = -b * omega**2 * np.cos(omega * t0)
    scale_a = 0.3  # масштаб для наглядности
    ax1.arrow(x0, y0, scale_a*ax0, scale_a*ay0, head_width=0.05, head_length=0.1, 
              fc='purple', ec='purple', width=0.008, label='Ускорение a')
    ax1.text(x0 + scale_a*ax0/2, y0 + scale_a*ay0/2, 'a', 
             ha='center', va='center', color='purple', fontsize=12)
    
    # Начало координат
    ax1.plot(0, 0, 'ko', markersize=8, label='Начало координат')
    ax1.plot(x0, y0, 'ro', markersize=6, label=f'Частица (t={t0/np.pi:.2f}π/ω)')
    
    # Подписи осей
    ax1.set_xlabel('x = a·sin(ωt)')
    ax1.set_ylabel('y = b·cos(ωt)')
    
    # Правая панель: зависимость силы от времени
    ax2.set_xlim(0, 4*np.pi/omega)
    ax2.grid(True, alpha=0.3)
    ax2.set_title('Зависимость модуля силы от времени')
    ax2.set_xlabel('Время t')
    ax2.set_ylabel('Модуль силы |F| / (mω²)')
    
    # Вычисляем модуль силы
    t_plot = np.linspace(0, 4*np.pi/omega, 1000)
    x_plot = a * np.sin(omega * t_plot)
    y_plot = b * np.cos(omega * t_plot)
    F_magnitude = np.sqrt((a * omega**2 * np.sin(omega * t_plot))**2 + 
                          (b * omega**2 * np.cos(omega * t_plot))**2) / (m * omega**2)
    
    # Нормализованный модуль силы (деленный на mω²)
    F_norm = np.sqrt(a**2 * np.sin(omega * t_plot)**2 + b**2 * np.cos(omega * t_plot)**2)
    
    ax2.plot(t_plot, F_norm, 'b-', label='|F|/(mω²)')
    ax2.axvline(x=t0, color='red', linestyle='--', alpha=0.7, 
                label=f'Момент t={t0/np.pi:.2f}π/ω')
    
    # Подпись на графике
    ax2.text(0.5, 0.95, f'|F| = mω²√(a²sin²(ωt) + b²cos²(ωt))', 
             transform=ax2.transAxes, fontsize=10, 
             bbox=dict(boxstyle="round,pad=0.3", facecolor="lightblue", alpha=0.5))
    
    # Масштаб для силы в выбранный момент
    F0_norm = np.sqrt(a**2 * np.sin(omega * t0)**2 + b**2 * np.cos(omega * t0)**2)
    ax2.plot(t0, F0_norm, 'ro', markersize=8)
    ax2.text(t0 + 0.1, F0_norm, f'  |F|/(mω²) = {F0_norm:.2f}', 
             ha='left', va='center', color='red')
    
    ax1.legend(loc='upper right')
    ax2.legend(loc='upper right')
    
    plt.tight_layout()
    plt.savefig('problem_1_95.png', dpi=300, bbox_inches='tight')
    plt.show()

def create_figure_1_8_simple():
    """Упрощенная версия рисунка для задачи 1.8"""
    fig = plt.figure(figsize=(10, 8))
    
    # Основная схема
    ax = plt.subplot(111)
    ax.set_xlim(-2, 2)
    ax.set_ylim(-2, 2)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.3)
    ax.set_title('Задача 1.8: Движение лодок на реке')
    
    # Река
    river = Rectangle((-1.8, -1.8), 3.6, 3.6, 
                      facecolor='lightblue', alpha=0.2, edgecolor='blue', linewidth=2)
    ax.add_patch(river)
    
    # Бакен в центре
    ax.plot(0, 0, 'ko', markersize=15, label='Бакен')
    ax.plot(0, 0, 'yo', markersize=8)
    
    # Стрелка течения
    ax.arrow(-1.5, 1.5, 2.8, 0, head_width=0.1, head_length=0.2, 
             fc='blue', ec='blue', width=0.03, linewidth=2)
    ax.text(0, 1.7, 'Течение реки: u', ha='center', va='center', 
            color='blue', fontsize=12, fontweight='bold')
    
    # Система координат
    ax.arrow(0, 0, 1.5, 0, head_width=0.08, head_length=0.15, 
             fc='black', ec='black', width=0.02, linewidth=1.5)
    ax.arrow(0, 0, 0, 1.5, head_width=0.08, head_length=0.15, 
             fc='black', ec='black', width=0.02, linewidth=1.5)
    ax.text(1.55, 0, 'x (вдоль реки)', ha='left', va='center', fontsize=11)
    ax.text(0, 1.55, 'y (поперёк реки)', ha='center', va='bottom', fontsize=11)
    
    # Лодка A - вдоль реки
    ax.arrow(0, 0.3, 1.2, 0, head_width=0.08, head_length=0.15, 
             fc='red', ec='red', width=0.02, linewidth=2)
    ax.arrow(0, 0.3, -1.2, 0, head_width=0.08, head_length=0.15, 
             fc='red', ec='red', width=0.02, linewidth=2, linestyle='--')
    ax.text(1.3, 0.3, 'Лодка A', ha='left', va='center', 
            color='red', fontsize=12, fontweight='bold')
    
    # Лодка B - поперёк реки
    ax.arrow(0.3, 0, 0, 1.2, head_width=0.08, head_length=0.15, 
             fc='green', ec='green', width=0.02, linewidth=2)
    ax.arrow(0.3, 0, 0, -1.2, head_width=0.08, head_length=0.15, 
             fc='green', ec='green', width=0.02, linewidth=2, linestyle='--')
    ax.text(0.3, 1.3, 'Лодка B', ha='left', va='center', 
            color='green', fontsize=12, fontweight='bold')
    
    # Вставка с формулами
    formula_text = (
        'Дано:\n'
        '• η = v/u = 1.2\n'
        '• u - скорость течения\n'
        '• v - скорость лодки отн. воды\n\n'
        'Для лодки A:\n'
        'τ_A = L/(v+u) + L/(v-u)\n\n'
        'Для лодки B:\n'
        'v_y = √(v² - u²)\n'
        'τ_B = 2L/v_y\n\n'
        'Ответ:\n'
        'τ_A/τ_B = η/√(η²-1) ≈ 1.81'
    )
    
    ax.text(1.6, 0.8, formula_text, ha='left', va='top', 
            fontsize=10, bbox=dict(boxstyle="round,pad=0.5", 
                                   facecolor="lightyellow", alpha=0.8))
    
    plt.tight_layout()
    plt.savefig('problem_1_8_simple.png', dpi=300, bbox_inches='tight')
    plt.show()

if __name__ == "__main__":
    # Масса для задачи 1.95 (произвольная)
    m = 1.0
    
    print("Создание рисунков для задач...")
    create_figure_1_8()
    print("Рисунок для задачи 1.8 создан: problem_1_8.png")
    
    create_figure_1_95()
    print("Рисунок для задачи 1.95 создан: problem_1_95.png")
    
    create_figure_1_8_simple()
    print("Упрощенный рисунок для задачи 1.8 создан: problem_1_8_simple.png")
    
    print("\nФайлы сохранены в текущей директории.")
```
