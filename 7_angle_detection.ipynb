```python
import cv2 
import numpy as np

# Загрузка изображения
image = cv2.imread("5table.jpg")
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
gray = np.float32(gray)

# Применение детектора Харриса
harris_response = cv2.cornerHarris(
    gray,
    blockSize=2,
    ksize=3,
    k=0.04
)

# Выделение углов на изображении
image[harris_response > 0.6 * harris_response.max()] = [0, 0, 255]

cv2.imshow("Harris Corners", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# Загрузка изображения
image = cv2.imread("image.jpg")
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Поиск углов
corners = cv2.goodFeaturesToTrack(
    gray,
    maxCorners=100,
    qualityLevel=0.01,
    minDistance=10
)

# Отрисовка углов
if corners is not None:
    corners = np.int32(corners)
    for corner in corners:
        x, y = corner.ravel()
        cv2.circle(image, (x, y), 4, (0, 255, 0), -1)

cv2.imshow("Shi-Tomasi Corners", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

image = cv2.imread("7table.jpg")        
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# 2. Поиск углов (Shi–Tomasi)
corners = cv2.goodFeaturesToTrack(
    gray,
    maxCorners=4,        # нам нужно ровно 4 угла
    qualityLevel=0.01,
    minDistance=50
)

# 3. Приводим координаты к int
corners = np.int32(corners)

# 4. Рисуем найденные углы
points = []
for c in corners:
    x, y = c.ravel()
    points.append((x, y))
    cv2.circle(image, (x, y), 6, (0, 0, 255), -1)

# 5. Показ результата
cv2.imshow("Desk corners", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
def polygon_area(points): #Используем формулу Гаусса
    area = 0
    n = len(points)
    for i in range(n):
        x1, y1 = points[i]
        x2, y2 = points[(i + 1) % n]
        area += x1 * y2 - y1 * x2
    return abs(area) / 2

area_pixels = polygon_area(points)
print("Площадь стола:", area_pixels, "пикселей²")
```
