```python
import cv2
import numpy as np


def process_apple_center_mass(path):
    # Загружаем изображение
    img = cv2.imread(path)

    # Переводим в HSV
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

    # Подходящий диапазон зелёного яблока
    hsv_min = np.array((35, 40, 40))
    hsv_max = np.array((90, 255, 255))

    # Маска
    mask = cv2.inRange(hsv, hsv_min, hsv_max)

    # Морфология (убрать дыры и шум)
    kernel = np.ones((7, 7), np.uint8)
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)

    # Центр масс 
    M = cv2.moments(mask)

    cx = int(M["m10"] / M["m00"])
    cy = int(M["m01"] / M["m00"])

    center = (cx, cy)

    #Радиус как max расстояние от центра до белой точки 
    ys, xs = np.where(mask == 255)

    distances = np.sqrt((xs - cx)**2 + (ys - cy)**2)
    r = int(np.max(distances))

    # Рисуем 
    out = img.copy()

    # окружность вокруг яблока
    cv2.circle(out, center, r, (0, 0, 255), 3)

    # центр масс
    cv2.circle(out, center, 6, (255, 0, 0), -1)

    

    return out, mask




result, mask = process_apple_center_mass("apple.png")

cv2.imshow("Apple via Mass Center", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
