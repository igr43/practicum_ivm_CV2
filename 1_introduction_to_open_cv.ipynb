```python
def resize_with_aspect_ratio(image, width=None, height=None, inter=cv2.INTER_AREA):
    dim = None
    (h, w) = image.shape[:2]
    
    if width is None and height is None:
        return image
        
    if width is None:
        r = height / float(h)
        dim = (int(w * r), height)
    else:
        r = width / float(w)
        dim = (width, int(h * r))
        
    return cv2.resize(image, dim, interpolation=inter)

# Пример использования
resized = resize_with_aspect_ratio(image, width=500)  # Меняем ширину, сохраняя пропорции

cv2.imshow('resized', resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2


def resize_with_aspect_ratio(image, width=None, height=None, inter=cv2.INTER_AREA):
    dim = None
    (h, w) = image.shape[:2]

    if width is None and height is None:
        return image

    if width is None:
        r = height / float(h)
        dim = (int(w * r), height)
    else:
        r = width / float(w)
        dim = (width, int(h * r))

    return cv2.resize(image, dim, interpolation=inter)



image = cv2.imread('images/balloon.jpg')


# 2. Примерный ROI для правого верхнего угла
h, w = image.shape[:2]

# Примерные координаты
x = int(w * 0.52)     
y = int(h * 0.27)     
roi_w = int(w * 0.35) # ширина прямоугольника
roi_h = int(h * 0.40) # высота прямоугольника 

cropped = image[y:y+roi_h, x:x+roi_w]


# 3. Изменяем размер, сохраняя пропорции
#resized = resize_with_aspect_ratio(cropped, width=600)
stretched = cv2.resize(cropped, (h, w))

cv2.imshow("Original", image)
cv2.imshow("Balloon ROI", stretched)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
